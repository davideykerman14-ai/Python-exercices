```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")
taxis = sns.load_dataset("taxis")

plt.figure(figsize=(8, 5))

sns.histplot(data=titanic, x="age", bins=30)

plt.title("Répartition des âges des passagers")
plt.xlabel("Âge (années)")
plt.ylabel("Nombre de passagers")

plt.savefig("figures/01_ages.png", dpi=150, bbox_inches="tight")
plt.show()

plt.figure(figsize=(8, 5))

sns.barplot(data=titanic, x="class", y="survived", hue="sex")

plt.title("Taux de survie selon la classe et le sexe")
plt.xlabel("Classe du billet")
plt.ylabel("Taux de survie")
plt.ylim(0, 1)

plt.savefig("figures/02_survie.png", dpi=150, bbox_inches="tight")
plt.show()

plt.figure(figsize=(8, 5))

sns.boxplot(data=titanic, x="class", y="fare")

plt.title("Prix du billet selon la classe")
plt.xlabel("Classe du billet")
plt.ylabel("Prix du billet (£)")

plt.savefig("figures/03_prix_classe.png", dpi=150, bbox_inches="tight")
plt.show()

taxis["pickup"] = pd.to_datetime(taxis["pickup"])

taxis["jour"] = taxis["pickup"].dt.date

ca_par_jour = taxis.groupby("jour")["total"].sum()

plt.figure(figsize=(10, 5))

sns.lineplot(x=ca_par_jour.index, y=ca_par_jour.values)

plt.title("Chiffre d'affaires des taxis par jour")
plt.xlabel("Jour")
plt.ylabel("Chiffre d'affaires ($)")
plt.xticks(rotation=45)

plt.savefig("figures/04_ca_journalier.png", dpi=150, bbox_inches="tight")
plt.show()

plt.figure(figsize=(8, 5))

sns.scatterplot(data=taxis, x="distance", y="total", alpha=0.5)

plt.title("Relation entre la distance et le prix de la course")
plt.xlabel("Distance (miles)")
plt.ylabel("Prix total ($)")

plt.savefig("figures/05_distance_prix.png", dpi=150, bbox_inches="tight")
plt.show()




```
