```python
import seaborn as sns

titanic = sns.load_dataset("titanic")

import pandas as pd

print(titanic.isna().sum())

mediane_age = titanic.groupby(["pclass", "sex"])["age"].transform("median")

nb_ages_manquants = titanic["age"].isna().sum()

titanic["age"] = titanic["age"].fillna(mediane_age)

print("Âges remplacés :", nb_ages_manquants)

titanic["embark_town"] = titanic["embark_town"].fillna("Inconnu")

"""titanic (titanic["age"] > 15 )"""

print("Valeurs manquantes dans deck :", titanic["deck"].isna().sum())

nb_lignes = len(titanic)

titanic = titanic.drop(columns=["deck"])

print("Colonne deck supprimée :", nb_lignes, "lignes conservées")

print("Doublons :", titanic.duplicated().sum())

# titanic = titanic.drop_duplicates()

titanic["fare"] = pd.to_numeric(titanic["fare"], errors="coerce")

print(titanic.dtypes)

q1 = titanic["fare"].quantile(0.25)
q3 = titanic["fare"].quantile(0.75)

ecart = q3 - q1
borne_haute = q3 + 1.5 * ecart

billets_atypiques = titanic[titanic["fare"] > borne_haute]

print("Borne haute :", borne_haute)
print(billets_atypiques[["fare"]])

print("Valeurs manquantes restantes :")
print(titanic.isna().sum())

titanic.to_csv("titanic_propre.csv", index=False)

print("Fichier enregistré : data/titanic_propre.csv")
```
