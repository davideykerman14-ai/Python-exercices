```python
import pandas as pd
import seaborn as sns

taxis = sns.load_dataset("taxis")

taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["dropoff"] = pd.to_datetime(taxis["dropoff"])

print(taxis.shape)
print(taxis.head())


taxis["jour"] = taxis["pickup"].dt.date

courses_par_jour = taxis.groupby("jour").agg(
nombre_courses=("pickup", "count"),
chiffre_affaires=("total", "sum")
)

print(courses_par_jour.head())

taxis["heure"] = taxis["pickup"].dt.hour

courses_par_heure = taxis.groupby("heure").size()

print(courses_par_heure)
print("Heure la plus chargée :", courses_par_heure.idxmax(), "h")

taxis["jour_semaine"] = taxis["pickup"].dt.day_name()

pourboires = taxis.groupby("jour_semaine")["tip"].mean()

print(pourboires)
print("Jour avec le pourboire moyen le plus élevé :", pourboires.idxmax())

taxis["duree_minutes"] = (
taxis["dropoff"] - taxis["pickup"]
).dt.total_seconds() / 60

print(taxis["duree_minutes"].describe())

print("Durées négatives :", (taxis["duree_minutes"] < 0).sum())
print("Durées supérieures à 3 heures :", (taxis["duree_minutes"] > 180).sum())

taxis["duree_verifiee"] = taxis["duree_minutes"].where(
(taxis["duree_minutes"] >= 0) &
(taxis["duree_minutes"] <= 180)
)

print(taxis[["duree_minutes", "duree_verifiee"]].head())





```
