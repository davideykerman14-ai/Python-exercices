```python
import seaborn as sns

titanic = sns.load_dataset("titanic")

mineurs = titanic[titanic["age"] < 18]
print(mineurs)

billets_les_chers = titanic.sort_values("fare", ascending=False)
print(billets_les_chers)

femmes_1ere = titanic[(titanic["sex"] == "female") & (titanic["class"] == "First")]
print(femmes_1ere) 

survie_femmes = femmes_1ere["survived"].mean() * 100
print(survie_femmes)

homme_3eme = titanic[(titanic["sex"] == "male") & (titanic["class"] == "Third")]
print(homme_3eme)

homme_3eme = homme_3eme["survived"].mean() * 100
print(homme_3eme)
```
